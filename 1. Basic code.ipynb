## 모듈 불러오기

```python
from matplotlib import pyplot as plt
from PIL import Image # PIL : Python Imaging Library
import cv2
import numpy as np
```

## Array 를 이미지로 만들기

```python
arr = np.arange(4).reshape((2,2))
```

```python
arr
```

```python
plt.imshow(arr, vmin=0, cmap='gray')
plt.grid(False)
plt.show()
```

```python
eye
```

## 이미지 불러오기

```python
path = './data/dog.jpg'
```

```python
cimg = Image.open(path)
cimg.show()
```

```python
dog_img = cv2.imread(path, cv2.IMREAD_COLOR)
plt.imshow(dog_img)
plt.show()
```

```python
dog_rgb = cv2.cvtColor(dog_img, cv2.COLOR_BGR2RGB)
plt.imshow(dog_rgb)
plt.show()
```

```python
dog_img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
```

```python
plt.imshow(dog_img, vmin=0, vmax=255, cmap='gray')
plt.show()
```

```python
dog_img
```

```python
dog_img.shape
```

## 이미지 변환하기

```python
plt.imshow(dog_img[100:200,100:200], vmin=0, vmax=255, cmap='gray')
plt.show()
```

```python
plt.imshow(dog_img /2, vmin=0, vmax=255, cmap='gray')
plt.show()
```

## 필터 씌우기

```python
arr = np.empty(350 * 350)

for idx in range(350 * 350):
  if idx % 2 == 0 :
    arr[idx] = 0
  else:
    arr[idx] = 1
```

```python
arr = arr.reshape((350, 350))
arr
```

```python
plt.imshow(arr, vmin=0, vmax=1, cmap='gray')
plt.show()
```

```python
plt.imshow(dog_img * arr, vmin=0, vmax=255, cmap='gray')
plt.show()
```

## Window WW/WL 조절하기

- Lung Window : 1500 / -700

- Mediastinum Window : 400 / 30

```python
print(dcm.WindowWidth, dcm.WindowCenter)
```

```python
ww = 1500
wc = -700
dcm.WindowCenter = wc
dcm.WindowWidth = ww
image = apply_modality_lut(dcm_img, dcm)
image_wwl = np.clip(image, wc - (ww / 2), wc + (ww / 2))
plt.imshow(image_wwl, cmap='gray')
plt.show()
```
